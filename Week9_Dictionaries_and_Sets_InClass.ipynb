## 1. What Is a Dictionary?

A **dictionary** stores a collection of **key-value pairs**. It *maps* each key to a value.

```python
dictionary = {key1: value1, key2: value2}
```

| | List | Dictionary |
|---|---|---|
| Syntax | `["Latte", "Mocha"]` | `{"Latte": 4.50, "Mocha": 4.75}` |
| Look up by | position (index) | key |
| Mutable? | ✅ yes | ✅ yes |

**Keys must be immutable** (strings, numbers, tuples). Values can be anything.

````python
menu = {"Latte": 4.50, "Mocha": 4.75, "Cold Brew": 5.00, "Matcha": 5.25}
phonebook = {"Chris": "555-1111", "Katie": "555-2222", "Joanne": "555-3333"}

print(menu)
print(phonebook)
````

## 2. Retrieving a Value

Use `dictionary[key]`. If the key isn't in the dictionary, Python raises a `KeyError`.

````python
print(menu["Mocha"])
print(phonebook["Katie"])
````

````python
try:
    print(menu["Americano"])
except KeyError as err:
    print("KeyError:", err)
````

Use `in` / `not in` to check for a key **before** you look it up. This prevents `KeyError`.

⚠️ `in` checks the **keys**, not the values.

````python
drink = "Americano"
if drink in menu:
    print(f"{drink} costs ${menu[drink]:.2f}")
else:
    print(f"Sorry, we don't sell {drink}.")

print("Latte" in menu)   # True  (a key)
print(4.50 in menu)      # False (a value, not a key)
````

## 3. Adding, Changing and Deleting

Dictionaries are **mutable**.

- `dictionary[key] = value` **adds** the pair if the key is new, or **changes** the value if the key already exists.
- `del dictionary[key]` deletes the pair (`KeyError` if the key is missing).
- `len(dictionary)` gives the number of pairs.

````python
menu = {"Latte": 4.50, "Mocha": 4.75, "Cold Brew": 5.00, "Matcha": 5.25}

menu["Espresso"] = 3.50     # add (new key)
menu["Latte"] = 4.75        # change (existing key)
del menu["Cold Brew"]       # delete

print(menu)
print("Number of drinks:", len(menu))
````

### Mixing data types

A single dictionary can use keys of different immutable types, and its values can be any type, even lists.

````python
student = {
    "name": "Ralphie",
    "id": 1001,
    "scores": [88, 92, 79],      # a list as a value
    (40.0, -105.3): "Boulder",   # a tuple as a key
}
print(student["scores"][1])
print(student[(40.0, -105.3)])
````

## 4. Empty Dictionaries and `for` Loops

Create an empty dictionary with `{}` or `dict()`, then add pairs as the program runs.

A `for` loop over a dictionary gives you its **keys**: `for key in dictionary:`

````python
inventory = {}            # or dict()
inventory["skis"] = 12
inventory["boots"] = 20
inventory["helmets"] = 8

for item in inventory:
    print(f"{item}: {inventory[item]}")
````

````python
# Building a dictionary in a loop: count how many times each word appears
words = ["latte", "mocha", "latte", "chai", "latte", "mocha"]

counts = {}
for w in words:
    if w in counts:
        counts[w] += 1
    else:
        counts[w] = 1
print(counts)
````

## 5. Dictionary Methods

| Method | What it does |
|---|---|
| `get(key, default)` | returns the value, or `default` if the key is missing (never raises `KeyError`) |
| `items()` | all key-value pairs, as tuples |
| `keys()` | all keys |
| `values()` | all values |
| `pop(key, default)` | removes the key and **returns** its value (`default` if missing) |
| `popitem()` | removes and returns the **last added** pair, as a tuple |
| `clear()` | deletes every pair, leaving `{}` |

````python
menu = {"Latte": 4.50, "Mocha": 4.75, "Matcha": 5.25, "Espresso": 3.50}

# get(): a safe lookup
print(menu.get("Mocha", "Not Available"))
print(menu.get("Americano", "Not Available"))

# keys() and values()
print(list(menu.keys()))
print(list(menu.values()))
print("Average price:", round(sum(menu.values()) / len(menu), 2))
````

`items()` gives `(key, value)` tuples. Unpack them into two loop variables:

````python
for drink, price in menu.items():
    print(f"{drink:<10} ${price:.2f}")
````

````python
removed = menu.pop("Matcha")
print("pop removed:", removed, "->", menu)

print("pop with default:", menu.pop("Chai", "not on menu"))

last = menu.popitem()
print("popitem removed:", last, "->", menu)

menu.clear()
print("after clear:", menu)
````

## 6. Common Loop Patterns

These three come up constantly: **total**, **update every value**, and **find the max**.

````python
rentals = {"Skis": 45, "Boots": 30, "Helmet": 15, "Poles": 10}
order = ["Skis", "Boots", "Helmet"]

# Total cost of an order
total = 0
for item in order:
    total += rentals[item]
print(f"Total rental cost: ${total}")

# Apply a 10% discount to every value
for item in rentals:
    rentals[item] = round(rentals[item] * 0.9, 2)
print(rentals)

# Find the key with the highest value
lift_tickets = {"Eldora": 129, "Copper Mountain": 159, "Breckenridge": 179, "Vail": 229}
best = None
for resort, price in lift_tickets.items():
    if best is None or price > lift_tickets[best]:
        best = resort
print(f"Most expensive: {best} at ${lift_tickets[best]}")
````

## 7. Merging: `|` and `|=`

- `dict3 = dict1 | dict2` builds a **new** dictionary with all pairs from both.
- `dict1 |= dict2` merges `dict2` **into** `dict1`.

If a key is in both, the value from the **right-hand** dictionary wins.

````python
dict1 = {"x": 1, "y": 2}
dict2 = {"y": 99, "z": 3}

dict3 = dict1 | dict2
print("dict1 | dict2 =", dict3)
print("dict1 unchanged:", dict1)

dict1 |= dict2
print("after dict1 |= dict2:", dict1)
````

## 8. Dictionary Comprehensions

A **dictionary comprehension** builds a dictionary in one line:

```python
new_dict = {key_expr: value_expr  for item in sequence  if condition}
```

The two cells below produce the same result:

````python
numbers = [1, 2, 3, 4]

# for loop version
squares = {}
for item in numbers:
    squares[item] = item ** 2
print(squares)

# dictionary comprehension version
squares = {item: item ** 2 for item in numbers}
print(squares)
````

````python
# Keys from a list, values computed from the key
names = ["Jeremy", "Kate", "Peg"]
str_lengths = {name: len(name) for name in names}
print(str_lengths)
````

### Comprehensions over `.items()`

Use `for k, v in d.items()` when you need both the key and the value. With `{k: v ...}` you get a copy of the dictionary.

````python
dict1 = {"A": 1, "B": 2, "C": 3}

copy = {k: v for k, v in dict1.items()}
doubled = {k: v * 2 for k, v in dict1.items()}
print(copy)
print(doubled)
````

## 9. Sets

A **set** works like a mathematical set:
- **No duplicates**: adding an item twice keeps only one.
- **No order**: there's no index, so `my_set[0]` doesn't work.
- **Mixed types** are allowed.
- **Fast membership checks**: `x in my_set` is much faster than searching a list.

Create a set with `set()`. ⚠️ `{}` makes an empty **dictionary**, not an empty set.

````python
empty = set()
print(type(empty), type({}))

colors = set(["red", "green", "red", "blue"])   # duplicates are dropped
print(colors)

letters = set("banana")                         # a string becomes its characters
print(letters)

also_a_set = {1, 2, 3}                          # literal with values works too
print(also_a_set, len(also_a_set))
````

### Removing duplicates from a list

This is one of the most common uses of a set. Convert to a set, then back to a list.

````python
customer_ids = [101, 205, 101, 330, 205, 101, 412]
unique_ids = list(set(customer_ids))
print(sorted(unique_ids))
print(f"{len(customer_ids)} visits from {len(unique_ids)} unique customers")
````

## 10. Adding and Removing Set Elements

| Method | What it does |
|---|---|
| `add(item)` | adds one element |
| `update(sequence)` | adds every element of a sequence |
| `remove(item)` | removes `item`, **`KeyError` if missing** |
| `discard(item)` | removes `item`, **no error** if missing |
| `clear()` | removes everything |

````python
tags = {"python", "data"}

tags.add("ai")
tags.add("python")            # already there: no change
tags.update(["sql", "excel"])
print(tags)

tags.remove("excel")
tags.discard("java")          # not there, but no error
print(tags)

try:
    tags.remove("java")
except KeyError as err:
    print("KeyError:", err)
````

## 11. `for`, `in` and `not in` With Sets

````python
approved_vendors = {"Acme", "Globex", "Initech"}

for vendor in approved_vendors:
    print(vendor)

invoice_vendor = "Umbrella"
if invoice_vendor not in approved_vendors:
    print(f"⚠️ {invoice_vendor} is not an approved vendor")
````

## 12. Set Operations

| Operation | Method | Operator | Result |
|---|---|---|---|
| Union | `a.union(b)` | `a \| b` | everything in either set |
| Intersection | `a.intersection(b)` | `a & b` | only what's in **both** |
| Difference | `a.difference(b)` | `a - b` | in `a` but **not** in `b` |
| Symmetric difference | `a.symmetric_difference(b)` | `a ^ b` | in one set or the other, **not both** |
| Subset | `a.issubset(b)` | `a <= b` | is every element of `a` in `b`? |
| Superset | `a.issuperset(b)` | `a >= b` | does `a` contain every element of `b`? |

````python
monday  = {"Ana", "Ben", "Cara", "Dev"}
tuesday = {"Cara", "Dev", "Eli", "Fay"}

print("Union:       ", monday | tuesday, monday.union(tuesday) == (monday | tuesday))
print("Intersection:", monday & tuesday)
print("Difference:  ", monday - tuesday)
print("Sym. diff:   ", monday ^ tuesday)
````

````python
set_a = {1, 2, 3, 4}
set_b = {2, 3}

print(set_b.issubset(set_a), set_b <= set_a)       # True True
print(set_a.issuperset(set_b), set_a >= set_b)     # True True
print(set_a.issubset(set_b))                       # False
````

## 13. Set Comprehensions

Same as a list comprehension, but with **curly braces** `{}`. The result has no duplicates.

````python
set1 = set([1, 2, 3, 4, 5])
print({item for item in set1})              # copy
print({item ** 2 for item in set1})         # squares

set2 = set([1, 20, 2, 40, 3, 50])
print({item for item in set2 if item < 10}) # filter

emails = ["ana@cu.edu", "ben@gmail.com", "cara@cu.edu", "dev@yahoo.com"]
domains = {e.split("@")[1] for e in emails}
print(domains)
````

## 14. Dictionaries + Sets Together

A dictionary comprehension can use a set as its input sequence.

````python
warehouse_a = {"laptop", "mouse", "keyboard"}
warehouse_b = {"keyboard", "monitor", "mouse"}
prices = {"laptop": 1200, "mouse": 25, "keyboard": 45, "monitor": 300}

common = warehouse_a & warehouse_b
common_prices = {item: prices[item] for item in common}
print(common)
print(common_prices)

favorite_lang = {"Ana": "Python", "Ben": "SQL", "Cara": "Python", "Dev": "R"}
print("Unique languages:", set(favorite_lang.values()))
````

## 15. Serializing Objects with `pickle`

**Serializing** turns an object into a stream of bytes so it can be saved to a file. In Python this is called **pickling**.

- Pickle: open the file for **binary writing** (`"wb"`) and call `pickle.dump(object, file)`.
- Unpickle: open the file for **binary reading** (`"rb"`) and call `pickle.load(file)`.

Unlike a text file, you get back the exact object, with its type, not a string you have to convert.

````python
import pickle

phonebook = {"Chris": "555-1111", "Katie": "555-2222", "Joanne": "555-3333"}

with open("phonebook.dat", "wb") as output_file:
    pickle.dump(phonebook, output_file)

with open("phonebook.dat", "rb") as input_file:
    pb = pickle.load(input_file)

print(pb)
print(type(pb))
````

````python
# You can pickle several objects into one file. Load them back in the same order.
customers = {"C1": "Ana", "C2": "Ben"}
vip_ids = {"C2"}

with open("store.dat", "wb") as f:
    pickle.dump(customers, f)
    pickle.dump(vip_ids, f)

with open("store.dat", "rb") as f:
    loaded_customers = pickle.load(f)
    loaded_vips = pickle.load(f)

print(loaded_customers, loaded_vips)
````

⚠️ Only unpickle files you trust. Loading a pickle can run code, so never open one from an unknown source.

## 16. Quick Recap

- Dictionary: `{key: value}`. Keys are unique and immutable. Look up with `d[key]`.
- `d[key] = value` adds or changes. `del d[key]` deletes. Missing keys raise `KeyError`.
- Use `in` or `d.get(key, default)` to avoid `KeyError`.
- `for k in d` loops over keys. `for k, v in d.items()` gives both.
- `keys()`, `values()`, `items()`, `pop()`, `popitem()`, `clear()`.
- `d1 | d2` merges into a new dictionary. `d1 |= d2` updates `d1`. The right side wins.
- Dictionary comprehension: `{k: v for k, v in d.items() if condition}`.
- Set: unique, unordered. Empty set is `set()`, **not** `{}`.
- `add`, `update`, `remove` (error if missing), `discard` (no error).
- `|` union, `&` intersection, `-` difference, `^` symmetric difference, `<=` subset, `>=` superset.
- Set comprehension: `{expr for item in seq if condition}`.
- `pickle.dump(obj, f)` with `"wb"`, `pickle.load(f)` with `"rb"`.
